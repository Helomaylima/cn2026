## Exercício 1 — Forma matricial $A\mathbf{x}=\mathbf{b}$

Cada linha do sistema vira uma linha de $A$ (coeficientes das incógnitas); o lado direito forma $\mathbf{b}$.

### a)
$$
\begin{cases} 2x + y = 7 \\ x - 3y = -5 \end{cases}
\quad\Longrightarrow\quad
\underbrace{\begin{bmatrix} 2 & 1 \\ 1 & -3 \end{bmatrix}}_{A}
\underbrace{\begin{bmatrix} x \\ y \end{bmatrix}}_{\mathbf{x}}
=
\underbrace{\begin{bmatrix} 7 \\ -5 \end{bmatrix}}_{\mathbf{b}}
$$

### b)
$$
\begin{cases} x + 2y - z = 2 \\ 3x - y + 2z = 9 \\ 2x + y + z = 7 \end{cases}
\quad\Longrightarrow\quad
\underbrace{\begin{bmatrix} 1 & 2 & -1 \\ 3 & -1 & 2 \\ 2 & 1 & 1 \end{bmatrix}}_{A}
\underbrace{\begin{bmatrix} x \\ y \\ z \end{bmatrix}}_{\mathbf{x}}
=
\underbrace{\begin{bmatrix} 2 \\ 9 \\ 7 \end{bmatrix}}_{\mathbf{b}}
$$


```python
# Exercício 1 — montagem das matrizes
A1a = np.array([[2, 1],
                [1, -3]], dtype=float)
b1a = np.array([7, -5], dtype=float)

A1b = np.array([[1, 2, -1],
                [3, -1, 2],
                [2, 1, 1]], dtype=float)
b1b = np.array([2, 9, 7], dtype=float)

print("1a) solução:", np.linalg.solve(A1a, b1a))
print("1b) solução:", np.linalg.solve(A1b, b1b))
```

Conferência: 1a) $x=\tfrac{16}{7}\approx 2{,}2857,\ y=\tfrac{17}{7}\approx 2{,}4286$; 1b) $x=2,\ y=1,\ z=2$.

## Exercício 2 — Eliminação de Gauss

$$
\begin{cases} 2x + y - z = 8 \\ -3x - y + 2z = -11 \\ -2x + y + 2z = -3 \end{cases}
$$

**Matriz aumentada inicial:**
$$
\left[\begin{array}{ccc|c} 2 & 1 & -1 & 8 \\ -3 & -1 & 2 & -11 \\ -2 & 1 & 2 & -3 \end{array}\right]
$$

### Etapa 1 — zerar a coluna 1 abaixo do pivô $a_{11}=2$
Multiplicadores: $m_{21} = \frac{-3}{2} = -1{,}5$ e $m_{31} = \frac{-2}{2} = -1$.

- $L_2 \leftarrow L_2 - m_{21}L_1 = L_2 + 1{,}5\,L_1$
- $L_3 \leftarrow L_3 - m_{31}L_1 = L_3 + L_1$

$$
\left[\begin{array}{ccc|c} 2 & 1 & -1 & 8 \\ 0 & 0{,}5 & 0{,}5 & 1 \\ 0 & 2 & 1 & 5 \end{array}\right]
$$

### Etapa 2 — zerar a coluna 2 abaixo do pivô $a_{22}=0{,}5$
Multiplicador: $m_{32} = \frac{2}{0{,}5} = 4$.

- $L_3 \leftarrow L_3 - 4\,L_2$

$$
\left[\begin{array}{ccc|c} 2 & 1 & -1 & 8 \\ 0 & 0{,}5 & 0{,}5 & 1 \\ 0 & 0 & -1 & 1 \end{array}\right]
$$

### Substituição regressiva
- $-z = 1 \Rightarrow z = -1$
- $0{,}5y + 0{,}5(-1) = 1 \Rightarrow 0{,}5y = 1{,}5 \Rightarrow y = 3$
- $2x + 3 - (-1) = 8 \Rightarrow 2x = 4 \Rightarrow x = 2$

$$\boxed{x = 2,\quad y = 3,\quad z = -1}$$

**Verificação:** $2(2)+3-(-1)=8$ ✔; $-3(2)-3+2(-1)=-11$ ✔; $-2(2)+3+2(-1)=-3$ ✔.

## Exercício 3 — Número de soluções

Critério (Teorema de Rouché–Capelli), com $n$ incógnitas:
- $\text{posto}(A) = \text{posto}([A|b]) = n$ → **solução única**;
- $\text{posto}(A) = \text{posto}([A|b]) < n$ → **infinitas soluções**;
- $\text{posto}(A) < \text{posto}([A|b])$ → **sem solução** (sistema inconsistente).

### a) $\begin{cases} x+y=3 \\ 2x+2y=6 \end{cases}$
$$
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 2 & 2 & 6 \end{array}\right]
\xrightarrow{L_2\leftarrow L_2-2L_1}
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 0 & 0 & 0 \end{array}\right]
$$
A segunda linha é $0=0$ (sempre verdadeira). $\text{posto}(A)=\text{posto}([A|b])=1<2$.
**Infinitas soluções**: $x = 3 - t,\ y = t,\ t\in\mathbb{R}$ (a variável $y$ é livre).

### b) $\begin{cases} x+y=3 \\ 2x+2y=7 \end{cases}$
$$
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 2 & 2 & 7 \end{array}\right]
\xrightarrow{L_2\leftarrow L_2-2L_1}
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 0 & 0 & 1 \end{array}\right]
$$
A segunda linha é $0=1$, absurdo. $\text{posto}(A)=1<\text{posto}([A|b])=2$.
**Nenhuma solução** (retas paralelas distintas).

### c) $\begin{cases} x+y=3 \\ 2x-y=0 \end{cases}$
$$
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 2 & -1 & 0 \end{array}\right]
\xrightarrow{L_2\leftarrow L_2-2L_1}
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 0 & -3 & -6 \end{array}\right]
$$
$-3y=-6\Rightarrow y=2$ e $x=3-2=1$. $\text{posto}(A)=\text{posto}([A|b])=2=n$.
**Solução única**: $(x,y)=(1,2)$ (retas concorrentes).

**Conferência computacional dos postos:**

```python
# Exercício 3 — conferência via postos (Rouché–Capelli)
sistemas = {
    "a)": (np.array([[1, 1], [2, 2]], float), np.array([3, 6], float)),
    "b)": (np.array([[1, 1], [2, 2]], float), np.array([3, 7], float)),
    "c)": (np.array([[1, 1], [2, -1]], float), np.array([3, 0], float)),
}

for nome, (A, b) in sistemas.items():
    n = A.shape[1]
    rA = np.linalg.matrix_rank(A)
    rAb = np.linalg.matrix_rank(np.column_stack([A, b]))
    if rA < rAb:
        res = "sem solução"
    elif rA == n:
        res = "solução única"
    else:
        res = "infinitas soluções"
    print(f"{nome} posto(A)={rA}, posto([A|b])={rAb}, n={n} -> {res}")

# Forma escalonada reduzida (para ver as linhas 0=0 / 0=1)
print()
for nome, (A, b) in sistemas.items():
    print(nome, "rref de [A|b]:", sp.Matrix(np.column_stack([A, b]).tolist()).rref()[0].tolist())
```

# Parte 3 — Implementação numérica em Python

## Exercício 4 — Eliminação de Gauss sem pivoteamento

**Algoritmo:**
1. Formar a matriz aumentada $[A|b]$.
2. Para cada coluna $k = 0,\dots,n-2$ (pivô $a_{kk}$): para cada linha $i>k$, calcular $m_{ik}=a_{ik}/a_{kk}$ e fazer $L_i \leftarrow L_i - m_{ik}L_k$.
3. Substituição regressiva: $x_i = \dfrac{b_i - \sum_{j>i} a_{ij}x_j}{a_{ii}}$.

Complexidade: $O(n^3)$ na eliminação e $O(n^2)$ na substituição.

```python
def gauss_sem_pivoteamento(A, b, verbose=False, tol=1e-12):
    # Resolve Ax = b por eliminação de Gauss SEM pivoteamento.
    # Parâmetros: A (n x n), b (n,), verbose (mostra cada etapa), tol (pivô ~ 0).
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float).reshape(-1)
    n = A.shape[0]

    if A.shape != (n, n) or b.size != n:
        raise ValueError("A deve ser quadrada n x n e b deve ter n componentes.")

    # 1) matriz aumentada [A | b]
    Ab = np.hstack([A, b.reshape(-1, 1)])
    if verbose:
        print("Matriz aumentada inicial:\n", Ab, "\n")

    # 2) eliminação: zera os elementos abaixo de cada pivô
    for k in range(n - 1):
        pivo = Ab[k, k]
        if abs(pivo) < tol:
            raise ZeroDivisionError(
                f"Pivô nulo (ou quase) em A[{k},{k}]. "
                "Sem pivoteamento o método falha; use pivoteamento parcial."
            )
        for i in range(k + 1, n):
            m = Ab[i, k] / pivo               # multiplicador m_ik
            Ab[i, k:] = Ab[i, k:] - m * Ab[k, k:]   # Li <- Li - m*Lk
            if verbose:
                print(f"  m{i+1}{k+1} = {m:.4f}  ->  L{i+1} <- L{i+1} - ({m:.4f})*L{k+1}")
        if verbose:
            print(f"\nApós etapa {k+1}:\n", Ab, "\n")

    if abs(Ab[n - 1, n - 1]) < tol:
        raise ZeroDivisionError("Pivô final nulo: sistema singular.")

    # 3) substituição regressiva
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        soma = np.dot(Ab[i, i + 1:n], x[i + 1:n])
        x[i] = (Ab[i, n] - soma) / Ab[i, i]

    return x
```

### Teste com o sistema do Exercício 2 (e comparação com `numpy.linalg.solve`)

```python
A2 = np.array([[ 2,  1, -1],
               [-3, -1,  2],
               [-2,  1,  2]], dtype=float)
b2 = np.array([8, -11, -3], dtype=float)

x_gauss = gauss_sem_pivoteamento(A2, b2, verbose=True)
x_numpy = np.linalg.solve(A2, b2)

print("Solução (Gauss implementado):", x_gauss)
print("Solução (numpy.linalg.solve):", x_numpy)
print("Diferença máxima |x_gauss - x_numpy| =", np.max(np.abs(x_gauss - x_numpy)))
print("Resíduo ||Ax - b||_inf =", np.linalg.norm(A2 @ x_gauss - b2, np.inf))
```

**Interpretação.** O código reproduz exatamente as etapas manuais do Exercício 2 (multiplicadores $-1{,}5$, $-1$ e $4$) e retorna $x=2,\ y=3,\ z=-1$. A diferença em relação ao `numpy.linalg.solve` é da ordem do erro de arredondamento ($\approx 10^{-15}$), e o resíduo é praticamente nulo. O `numpy` usa fatoração LU com **pivoteamento parcial** (LAPACK), mais robusta; neste exemplo os pivôs ($2$; $0{,}5$; $-1$) não são nulos, então a versão sem pivoteamento funciona bem.

### Limitação: por que o pivoteamento importa

```python
# Exemplo em que o pivoteamento é necessário: pivô a11 = 0
A_ruim = np.array([[0, 1],
                   [1, 1]], dtype=float)
b_ruim = np.array([1, 2], dtype=float)

try:
    gauss_sem_pivoteamento(A_ruim, b_ruim)
except ZeroDivisionError as e:
    print("Gauss sem pivoteamento falhou:", e)

print("numpy.linalg.solve resolve normalmente:", np.linalg.solve(A_ruim, b_ruim))

# Pivô muito pequeno: amplifica erros de arredondamento
eps = 1e-17
A_eps = np.array([[eps, 1], [1, 1]], dtype=float)
b_eps = np.array([1, 2], dtype=float)
print("\nPivô pequeno (1e-17):")
print("  sem pivoteamento:", gauss_sem_pivoteamento(A_eps, b_eps, tol=0), " (solução correta ≈ [1, 1])")
print("  numpy (com pivot.):", np.linalg.solve(A_eps, b_eps))
```

**Conclusão.** Pivô nulo causa divisão por zero; pivô muito pequeno gera multiplicadores enormes e perda de precisão (cancelamento catastrófico). Por isso, na prática, usa-se **pivoteamento parcial** (trocar linhas para que o pivô seja o maior elemento em módulo da coluna).

## Exercício 5 — Modelo de balanço térmico

$$
\begin{bmatrix} 4 & -1 & 0 \\ -1 & 4 & -1 \\ 0 & -1 & 3 \end{bmatrix}
\begin{bmatrix} T_1 \\ T_2 \\ T_3 \end{bmatrix}
=
\begin{bmatrix} 15 \\ 10 \\ 10 \end{bmatrix}
$$

### a) Resolução manual (Gauss)

$$
\left[\begin{array}{ccc|c} 4 & -1 & 0 & 15 \\ -1 & 4 & -1 & 10 \\ 0 & -1 & 3 & 10 \end{array}\right]
$$

**Etapa 1:** $m_{21} = \frac{-1}{4} = -0{,}25$; $m_{31}=0$ (já é zero).
$L_2 \leftarrow L_2 + 0{,}25\,L_1$: $\;4 - 0{,}25 = 3{,}75$; $\;-1+0 = -1$; $\;10+3{,}75=13{,}75$.

$$
\left[\begin{array}{ccc|c} 4 & -1 & 0 & 15 \\ 0 & 3{,}75 & -1 & 13{,}75 \\ 0 & -1 & 3 & 10 \end{array}\right]
$$

**Etapa 2:** $m_{32}=\frac{-1}{3{,}75}=-\frac{4}{15}$. $L_3\leftarrow L_3+\frac{4}{15}L_2$:
- $3 - \frac{4}{15} = \frac{41}{15}$
- $10 + \frac{4}{15}\cdot 13{,}75 = 10+\frac{11}{3} = \frac{41}{3}$

$$
\left[\begin{array}{ccc|c} 4 & -1 & 0 & 15 \\ 0 & 3{,}75 & -1 & 13{,}75 \\ 0 & 0 & \frac{41}{15} & \frac{41}{3} \end{array}\right]
$$

**Substituição regressiva:**
- $\frac{41}{15}T_3=\frac{41}{3}\Rightarrow T_3 = 5$
- $3{,}75\,T_2 - 5 = 13{,}75 \Rightarrow T_2 = \frac{18{,}75}{3{,}75} = 5$
- $4T_1 - 5 = 15 \Rightarrow T_1 = 5$

$$\boxed{T_1 = T_2 = T_3 = 5}$$

**Verificação:** $4(5)-5=15$ ✔; $-5+20-5=10$ ✔; $-5+15=10$ ✔.

### b) Implementação em Python

```python
A5 = np.array([[ 4, -1,  0],
               [-1,  4, -1],
               [ 0, -1,  3]], dtype=float)
b5 = np.array([15, 10, 10], dtype=float)

T = gauss_sem_pivoteamento(A5, b5, verbose=True)
T_np = np.linalg.solve(A5, b5)

print("T (Gauss implementado) =", T)
print("T (numpy.linalg.solve) =", T_np)
print("Resíduo ||AT - b||_inf =", np.linalg.norm(A5 @ T - b5, np.inf))
print("Número de condição cond(A) =", np.linalg.cond(A5))
print("Diagonalmente dominante?  ",
      all(abs(A5[i, i]) >= np.sum(np.abs(A5[i])) - abs(A5[i, i]) for i in range(3)))
```

**Interpretação.** A solução numérica coincide com a manual: $T_1=T_2=T_3=5$. A matriz é simétrica, tridiagonal e **diagonalmente dominante** (com dominância estrita nas linhas 1 e 3) e positiva definida, logo é não singular e todos os pivôs são não nulos — por isso a eliminação sem pivoteamento é estável. O número de condição baixo ($\approx 2$) indica um problema bem condicionado: pequenas perturbações nos dados causam pequenas variações nas temperaturas. Fisicamente, o resultado uniforme indica que o balanço entre as fontes ($b$) e as trocas de calor entre os nós leva a um equilíbrio com temperaturas iguais.
