**Resposta do grupo:**
> Seria útil para estimar a temperatura de uma cidade em um horário intermediário (por exemplo, às 14h30) quando só temos medições registradas em horários cheios (como 14h e 15h), ou para prever a posição de um veículo entre leituras de sensores GPS. A estimativa pode se tornar pouco confiável se ocorrerem variações bruscas ou eventos atípicos no intervalo não medido (como uma tempestade repentina ou uma frenagem brusca), se os pontos medidos tiverem erros/ruídos de leitura, ou se o intervalo de tempo entre as medições originais for grande demais.

## Seção 3: Primeiro exemplo (observar antes de formalizar)

**Descrevam a tendência dos pontos. A curva precisa ser reta? Há mais de uma curva que poderia passar pelos pontos?**

**Resposta do grupo:**
> Os pontos $(-2, 9)$, $(-1, 4)$, $(0, 1)$ e $(1, 0)$ mostram uma tendência descendente que vai desacelerando conforme $x$ aumenta. Existem infinitas curvas que poderiam passar por esses 4 pontos, porém existe apenas **um único polinômio de grau no máximo 3** que intercepta exatamente todos eles.

## Seção 4: Construindo o polinômio com uma ferramenta

**Registrem a expressão obtida e identifiquem seu grau. Se algum coeficiente for praticamente zero, indiquem isso.**

**Resposta do grupo:**
> A expressão obtida pelo NumPy foi:
> $$P(x) = 1.0 - 2.0x + 1.0x^2 + (7.4383 \times 10^{-16})x^3$$
> O polinômio é de **grau 2** (uma parábola), pois o coeficiente do termo cúbico $x^3$ é da ordem de $10^{-16}$, o que numericamente equivale a zero (trata-se apenas de um resíduo de arredondamento por precisão computacional). Simplificando, o polinômio resultante é:
> $$P(x) = x^2 - 2x + 1 = (x - 1)^2$$

2. O que essa conferência mostra sobre a curva nos nós?
3. O que ela **não** mostra sobre valores entre os nós ou fora do intervalo?

**Resposta do grupo:**
> 1. As diferenças calculadas pelo código (da ordem de $10^{-15}$ a $10^{-16}$) são virtualmente zero, decorrentes apenas da precisão de ponto flutuante do computador.
> 2. A conferência comprova matematicamente a propriedade da interpolação: a curva calculada passa **exatamente** sobre todos os nós informados.
> 3. Ela **não** garante que o polinômio represente o comportamento real do fenômeno entre os nós (onde pode haver oscilações indesejadas) nem fora do intervalo coberto (extrapolação), onde a curva pode crescer ou decrescer de forma irrealista.

## Seção 7: Segundo conjunto (dados de medição)

**Discussão: $x=1{,}5$ está entre nós? Qual a estimativa?**

**Resposta do grupo:**
> Sim, $x = 1{,}5$ está compreendido no intervalo coberto pelos nós $[0, 3]$, tratando-se portanto de uma **interpolação**. A estimativa calculada pelo polinômio de grau 3 para esse instante é $P(1{,}5) = 2{,}3125$.

## Seção 8: Experimento (alterar uma medição)

**Registrem: qual ponto foi alterado? O que aconteceu ao valor em $x=1{,}5$? A curva mudou apenas perto do ponto alterado? Por que pequenas incertezas de medição podem ser importantes?**

**Resposta do grupo:**
> Alteramos o ponto no nó $x = 2$, modificando sua ordenada de $y = 2{,}7$ para $y = 3{,}2$ (um acréscimo de $+0{,}5$).
> Com essa alteração, a estimativa no ponto intermediário $x = 1{,}5$ mudou de $2{,}3125$ para $2{,}59375$. A curva alterou seu formato globalmente, afetando não apenas a vizinhança imediata de $x = 2$, mas todo o intervalo de interpolação. Isso mostra que a interpolação polinomial global é altamente sensível aos dados: pequenas incertezas de medição em um único ponto se propagam e distorcem a estimativa em outros trechos da curva.

## Seção 10: Fenômeno de Runge (alerta importante)

**O que acontece perto das bordas quando aumentamos o número de pontos? O polinômio passa pelos pontos? Ele representa bem a função no meio? E perto dos extremos?**

**Resposta do grupo:**
> Quando aumentamos o número de pontos com espaçamento uniforme para interpolar a função de Runge $f(x) = \frac{1}{1 + 25x^2}$, o polinômio continua passando **exatamente** sobre todos os pontos amostrados. No entanto, perto das bordas do intervalo (próximo de $x = -1$ e $x = 1$), surgem **oscilações extremas e descontroladas** (Fenômeno de Runge). No centro do intervalo (ao redor de $x = 0$), o polinômio representa bem a função original, mas perto dos extremos o erro aumenta drasticamente, tornando o polinômio de alto grau inadequado para prever valores nas regiões periféricas.
